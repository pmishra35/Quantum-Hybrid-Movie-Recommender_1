# 05 - Hybrid Model

Combine collaborative and demographic signals with alpha.

```python
from pathlib import Path
import sys
sys.path.append(str(Path.cwd().parent))

from src.preprocessing import prepare_data
from src.hybrid_model import HybridRecommender

data = prepare_data(sample_users=300)
for alpha in [0.25, 0.50, 0.75, 1.00]:
    model = HybridRecommender(alpha=alpha).fit(data.ratings, data.users, data.movies)
    print(alpha, model.recommend(user_id=int(data.users.user_id.iloc[0]), top_n=3)[['title', 'predicted_rating']])
```
